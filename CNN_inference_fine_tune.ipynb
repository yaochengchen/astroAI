```python
import pickle
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from astropy.coordinates import SkyCoord
from gammapy.maps import WcsGeom
from sklearn.model_selection import train_test_split

# =========================
# 0. config
# =========================
MODEL_PATH = "/home/cyc/software/astroAI/models/crta_models/regressor_z20.h5"
DATA_PATH = "/home/cyc/software/astroAI/dataset/simulations_dataset_v1/dl3_z20_irf/z20_clean_test.pickle"
MANIFEST_PATH = "/home/cyc/software/astroAI/dataset/simulations_dataset_v1/dl3_z20_irf/output_pairs/manifest.csv"

N_SAMPLES = 2000
N_FINETUNE = 500
RANDOM_SEED = 42

PIX_SIZE = 0.025
WIDTH = 5.0
IMAGE_SIZE = 200

BATCH_SIZE = 16
EPOCHS = 60
LEARNING_RATE = 8e-6

# True: 只训练最后两层 Dense，比较稳
# False: 全模型都参与微调
FREEZE_FEATURE_EXTRACTOR = True


# =========================
# 1. helper function
# =========================
def get_true_pixel_from_manifest(row, pixsize=0.025, width=5.0):
    ra_src = row["source_ra"]
    dec_src = row["source_dec"]
    ra_pnt = row["point_ra"]
    dec_pnt = row["point_dec"]

    geom = WcsGeom.create(
        skydir=SkyCoord(ra_pnt, dec_pnt, unit="deg", frame="icrs"),
        binsz=pixsize,
        width=(width, width),
        frame="icrs",
        proj="CAR"
    )

    coord = SkyCoord(ra_src, dec_src, unit="deg", frame="icrs")
    x_true, y_true = geom.coord_to_pix({"skycoord": coord})[:2]

    x_true = float(np.asarray(x_true).squeeze())
    y_true = float(np.asarray(y_true).squeeze())

    return x_true, y_true


# =========================
# 2. load model
# =========================
model = tf.keras.models.load_model(MODEL_PATH, compile=False)
print("Regressor loaded")

print(type(model))
print("Before fine-tuning:")
model.summary()

# 可选：冻结前面的卷积/特征层，只训最后 Dense 部分
if FREEZE_FEATURE_EXTRACTOR:
    for layer in model.layers[:-2]:
        layer.trainable = False
    for layer in model.layers[-2:]:
        layer.trainable = True
    print("\nFreeze feature extractor, only fine-tune the last two layers.")
else:
    for layer in model.layers:
        layer.trainable = True
    print("\nFine-tune all layers.")

print("\nTrainable status:")
for i, layer in enumerate(model.layers):
    print(f"{i:2d} | {layer.name:20s} | trainable={layer.trainable}")


def weighted_mse(y_true, y_pred):
    dx = y_pred[:, 0] - y_true[:, 0]
    dy = y_pred[:, 1] - y_true[:, 1]

    loss_x = tf.square(dx)
    loss_y = tf.square(dy)

    return tf.reduce_mean(2.0 * loss_x + 1.0 * loss_y)  # 比如 x 权重大

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE),
    #loss="mse",
    loss=weighted_mse,
    metrics=["mae"]
)


# =========================
# 3. load dataset
# =========================
with open(DATA_PATH, "rb") as f:
    ds = pickle.load(f)

x_all = ds["DS1"]   # noisy map
print("Data shape:", x_all.shape)

manifest = pd.read_csv(MANIFEST_PATH)
manifest = manifest.sort_values("seed").reset_index(drop=True)
print("Manifest shape:", manifest.shape)


# =========================
# 4. build full dataset (X, y)
# =========================
n_samples = min(N_SAMPLES, len(x_all), len(manifest))
print(f"Using n_samples = {n_samples}")

X = x_all[:n_samples].reshape(-1, IMAGE_SIZE, IMAGE_SIZE, 1).astype("float32")

true_xy = []
seed_list = []

for i in range(n_samples):
    seed = i + 1

    row_match = manifest[manifest["seed"] == seed]
    if len(row_match) == 0:
        raise ValueError(f"seed={seed} not found in manifest")

    row = row_match.iloc[0]
    true_x, true_y = get_true_pixel_from_manifest(row, pixsize=PIX_SIZE, width=WIDTH)

    true_xy.append([true_x, true_y])
    seed_list.append(seed)

true_xy = np.array(true_xy, dtype="float32")

# 标签归一化到 [0,1]，因为你的模型输出看起来就是归一化坐标
y = true_xy / IMAGE_SIZE

print("X shape:", X.shape)
print("y shape:", y.shape)
print("First 5 labels (pixel):")
print(true_xy[:5])
print("First 5 labels (normalized):")
print(y[:5])


# =========================
# 5. randomly choose 200 for fine-tuning
# =========================
rng = np.random.default_rng(RANDOM_SEED)
all_indices = np.arange(n_samples)
finetune_indices = rng.choice(all_indices, size=N_FINETUNE, replace=False)
eval_indices = np.setdiff1d(all_indices, finetune_indices)

print(f"Fine-tune samples: {len(finetune_indices)}")
print(f"Evaluation samples: {len(eval_indices)}")

X_finetune = X[finetune_indices]
y_finetune = y[finetune_indices]

X_eval = X[eval_indices]
y_eval = y[eval_indices]
true_xy_eval = true_xy[eval_indices]
seed_eval = np.array(seed_list)[eval_indices]

# 再从 200 个微调样本里切一部分做 validation
X_train, X_val, y_train, y_val = train_test_split(
    X_finetune,
    y_finetune,
    test_size=0.2,
    random_state=RANDOM_SEED
)

print("X_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_eval :", X_eval.shape)


# =========================
# 6. fine-tune
# =========================
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=7,
        restore_best_weights=True
    )
]

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    callbacks=callbacks,
    verbose=1
)


# =========================
# 7. evaluate on the remaining samples
# =========================
pred_eval = model.predict(X_eval, verbose=1)

pred_x_list = []
pred_y_list = []
true_x_list = []
true_y_list = []
dx_list = []
dy_list = []
dist_list = []
seed_result_list = []

for i in range(len(X_eval)):
    pred_x = float(pred_eval[i, 0] * IMAGE_SIZE)
    pred_y = float(pred_eval[i, 1] * IMAGE_SIZE)

    true_x = float(true_xy_eval[i, 0])
    true_y = float(true_xy_eval[i, 1])

    dx = pred_x - true_x
    dy = pred_y - true_y
    dist = np.sqrt(dx**2 + dy**2)

    pred_x_list.append(pred_x)
    pred_y_list.append(pred_y)
    true_x_list.append(true_x)
    true_y_list.append(true_y)
    dx_list.append(dx)
    dy_list.append(dy)
    dist_list.append(dist)
    seed_result_list.append(int(seed_eval[i]))

    if i < 5:
        print(
            f"seed={int(seed_eval[i]):4d} | "
            f"pred=({pred_x:.2f}, {pred_y:.2f}) | "
            f"true=({true_x:.2f}, {true_y:.2f}) | "
            f"dx={dx:.2f}, dy={dy:.2f}, dist={dist:.2f}"
        )

result_df = pd.DataFrame({
    "seed": seed_result_list,
    "pred_x": pred_x_list,
    "pred_y": pred_y_list,
    "true_x": true_x_list,
    "true_y": true_y_list,
    "dx": dx_list,
    "dy": dy_list,
    "dist": dist_list,
})

print("\nResult head:")
print(result_df.head())

print("\nEvaluation summary:")
print(f"mean(dx)   = {np.mean(dx_list):.4f}")
print(f"std(dx)    = {np.std(dx_list):.4f}")
print(f"mean(dy)   = {np.mean(dy_list):.4f}")
print(f"std(dy)    = {np.std(dy_list):.4f}")
print(f"mean(dist) = {np.mean(dist_list):.4f}")
print(f"median(dist)= {np.median(dist_list):.4f}")


# =========================
# 8. save fine-tuned model and results
# =========================
model.save("/home/cyc/software/astroAI/models/crta_models/regressor_z20_finetuned.h5")
result_df.to_csv("finetune_eval_results.csv", index=False)

print("\nSaved fine-tuned model:")
print("/home/cyc/software/astroAI/models/crta_models/regressor_z20_finetuned.h5")
print("Saved evaluation table:")
print("finetune_eval_results.csv")


# =========================
# 9. plot training history
# =========================
plt.figure(figsize=(6, 5))
plt.plot(history.history["loss"], label="train_loss")
plt.plot(history.history["val_loss"], label="val_loss")
plt.xlabel("Epoch")
plt.ylabel("Loss (MSE)")
plt.title("Fine-tuning Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 5))
plt.plot(history.history["mae"], label="train_mae")
plt.plot(history.history["val_mae"], label="val_mae")
plt.xlabel("Epoch")
plt.ylabel("MAE")
plt.title("Fine-tuning MAE")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# =========================
# 10. plot evaluation results
# =========================
plt.figure(figsize=(6, 5))
plt.hist(dx_list, bins=50)
plt.xlabel("dx = pred_x - true_x [pixel]")
plt.ylabel("Count")
plt.title("Distribution of dx")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 5))
plt.hist(dy_list, bins=50)
plt.xlabel("dy = pred_y - true_y [pixel]")
plt.ylabel("Count")
plt.title("Distribution of dy")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 5))
plt.hist(dist_list, bins=50)
plt.xlabel("distance [pixel]")
plt.ylabel("Count")
plt.title("Distribution of prediction error distance")
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 6))
plt.scatter(dx_list, dy_list, s=8, alpha=0.6)
plt.xlabel("dx [pixel]")
plt.ylabel("dy [pixel]")
plt.title("dx vs dy")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from scipy.optimize import curve_fit

# 定义高斯函数
def gaussian(x, a, mu, sigma):
    return a * np.exp(-(x - mu)**2 / (2 * sigma**2))

def fit_and_plot_distribution(data, xlabel, title, filename):
    # 1. 计算直方图的 bin 和 counts
    counts, bin_edges = np.histogram(data, bins=50, density=True)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    
    # 2. 初始猜测值 (Initial guess)
    # A 猜测为最大值，mu 猜测为均值，sigma 猜测为标准差
    p0 = [max(counts), np.mean(data), np.std(data)]
    
    # 3. 进行高斯拟合
    try:
        popt, pcov = curve_fit(gaussian, bin_centers, counts, p0=p0)
        a_fit, mu_fit, sigma_fit = popt
        print(f"[{title}] Fit Result: mu={mu_fit:.4f}, sigma={sigma_fit:.4f}")
    except Exception as e:
        print(f"[{title}] Fitting failed: {e}")
        return

    # 4. 绘图
    plt.figure(figsize=(6, 5))
    plt.hist(data, bins=50, density=True, alpha=0.6, label="Data")
    
    # 绘制拟合曲线
    x_range = np.linspace(min(bin_centers), max(bin_centers), 100)
    plt.plot(x_range, gaussian(x_range, *popt), 'r-', lw=2, 
             label=f'Gaussian Fit:\n$\mu={mu_fit:.3f}, \sigma={sigma_fit:.3f}$')
    
    plt.xlabel(xlabel)
    plt.ylabel("Density")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()

# 分别对 dx, dy, dist 进行拟合
fit_and_plot_distribution(dx_list, "dx [pixel]", "Distribution of dx", "dx_dist_fit.png")
fit_and_plot_distribution(dy_list, "dy [pixel]", "Distribution of dy", "dy_dist_fit.png")
```
